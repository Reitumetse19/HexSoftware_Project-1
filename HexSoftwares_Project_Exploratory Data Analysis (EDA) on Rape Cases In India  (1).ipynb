## Data Distributions

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = depression_df_cleaned.select_dtypes(include=['number'])

sns.set(style="whitegrid")

for column in numeric_df.columns:
    plt.figure(figsize=(12, 5))

    
    plt.subplot(1, 2, 1)
    sns.histplot(numeric_df[column], kde=True, bins=30, color='purple')
    plt.title(f'Histogram of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')


    plt.subplot(1, 2, 2)
    sns.boxplot(x=numeric_df[column], color='c')
    plt.title(f'Boxplot of {column}')
    plt.xlabel(column)

    plt.tight_layout()
    plt.show()
```

## Basic Visualisation Charts and Graphs

```python
import matplotlib.pyplot as plt


avg_age_by_gender = depression_df_cleaned.groupby('Gender')['Age'].mean()
plt.figure(figsize=(8, 5))
avg_age_by_gender.plot(kind='bar', color=['skyblue', 'salmon'])

plt.title('Average Age by Gender')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

colors = {'Male': 'c', 'Female': 'm'}
for gender, color in colors.items():
    subset = depression_df_cleaned[depression_df_cleaned['Gender'] == gender]
    plt.scatter(subset['Gender'], subset['Age'], alpha=0.3, label=gender, color=color)

plt.title('Scatter Plot of Age by Gender')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(depression_df_cleaned['Age'], bins=20, color='green', edgecolor='yellow', alpha=0.7)

plt.title('Age Distribution of Respondents')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
degree_profession_counts = depression_df_cleaned.groupby('Degree')['Profession'].nunique()

plt.figure(figsize=(8, 8))
plt.pie(degree_profession_counts, labels=degree_profession_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Pie Chart of Professions by Degree')
plt.axis('equal') 
plt.tight_layout()
plt.show()
```

```python
age_range = [15, 20, 25, 30, 35, 40, 60]
ages = ['16-20', '21-25', '26-30', '31-35', '36-40', '41+']
depression_df_cleaned['Age Grouped'] = pd.cut(depression_df_cleaned['Age'], bins=age_range, labels=ages)

age_dietary = depression_df_cleaned.groupby(['Dietary Habits', 'Age Grouped']).size().unstack().fillna(0)
age_dietary.plot(kind='bar', stacked = True, figsize=(10, 6), colormap='viridis')
plt.title('Age Group Distribution by Dietary Habits')
plt.xlabel('Dietary Habits')
plt.ylabel('Number of Students')
plt.legend(title='Age Group')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

```

## Correlations

```python
numeric_df = depression_df_cleaned.select_dtypes(include=[np.number])

correlation = numeric_df.corr()
print(correlation)

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm", square=True, linewidths=0.5)
plt.title("Correlation Heatmap for the Depression Cases", fontsize=1)
plt.tight_layout()
plt.show()
```
